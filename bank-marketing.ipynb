# 3. Subscription Rate by Age Group   : Conversion rate across different age segments.  


```python
bins=[18,30,40,50,60,70,100]
labels = ["18-30","31-40","41-50","51-60","61-70","71-100"]
df["age_group"] = pd.cut(df["age"],bins=bins,labels=labels)
```

```python
df["age_group"].value_counts().sort_index()
```

```python
age_target = df.groupby("age_group")["y"].value_counts().sort_values(ascending = False).reset_index()
age_target
```

```python
fig_age_target=px.bar(age_target,
                     x="age_group",
                     y="count",
                     color="y",
                     title="Subscribtion by Age Group",
                     barmode="group",
                     text_auto = True)
fig_age_target.show()
```
